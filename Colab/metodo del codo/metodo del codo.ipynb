```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import pandas as pd
```

```python
data = {
    "Horas de estudio": [1,2,1.5,8,9,8.5,3,3.5,4],
    "Calificaciones": [2,3,2.5,9,10,9.5,5,5.5,6]
}
df=pd.DataFrame(data)
df
```

```python
plt.scatter(df["Horas de estudio"], df["Calificaciones"])
plt.xlabel("Horas de estudio")
plt.ylabel("Calificaciones")
plt.title("Datos de los estudiantes")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=0)
kmeans.fit(df)
```

```python
centroides = kmeans.cluster_centers_
centroides
```

```python
df['Cluster'] = kmeans.labels_
df
```

```python
plt.scatter(df["Horas de estudio"], df["Calificaciones"],c=df['Cluster'])
plt.scatter(centroides[:,0], centroides[:,1], marker="x",color="red",s=200, label="Centroides")
```

```python
# 1. Creamos una lista para ir guardando la inercia (WCSS) de cada modelo elaborado
inercia = []
# 2. Definimos que queremos probar dividiendo los datos desde 1 hasta 9 clusters
rango_k = range(1, 10)
# 3. Iteramos para cada K
for k in rango_k:
    # Creamos el modelo
    kmeans = KMeans(n_clusters=k, random_state=0, n_init='auto')
    
    # Entrenamos el modelo pasándole nuestro DataFrame
    kmeans.fit(df)
    
    # Guardamos el resultado de la inercia (Suma de las distancias al cuadrado)
    inercia.append(kmeans.inertia_)
# 4. Graficamos los resultados
plt.figure(figsize=(8, 5))
plt.plot(rango_k, inercia, marker='o', linestyle='-', color='b')
plt.title('Método del Codo para Encontrar el "K" Óptimo')
plt.xlabel('Número de clusters (k)')
plt.ylabel('Inercia (WCSS)')
plt.xticks(rango_k)
plt.grid(True)
plt.show()
```
