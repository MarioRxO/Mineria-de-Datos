```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

X = np.array([
    [95, 98],
    [92, 94],
    [88, 90],
    [65, 70],
    [60, 68],
    [58, 62],
    [30, 40],
    [35, 38]
])

labels = ['A','B','C','D','E','F','G','H']
```

```python
Z = linkage(X, method='ward')

plt.figure(figsize=(10,5))
dendrogram(Z, labels=labels)
plt.title("Dendrograma Rendimiento")
plt.show()
```

```python
modelo2 = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

c2 = modelo2.fit_predict(X)

plt.figure(figsize=(6,6))

for i in range(len(X)):
    plt.scatter(X[i,0], X[i,1], c=f"C{c2[i]}")
    plt.annotate(labels[i], (X[i,0], X[i,1]))

plt.title("2 Clusters")
plt.grid()
plt.show()
```

```python
modelo3 = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

c3 = modelo3.fit_predict(X)

plt.figure(figsize=(6,6))

for i in range(len(X)):
    plt.scatter(X[i,0], X[i,1], c=f"C{c3[i]}")
    plt.annotate(labels[i], (X[i,0], X[i,1]))

plt.title("3 Clusters")
plt.grid()
plt.show()
```

Los grupos se vuelven más generales o específicos.
