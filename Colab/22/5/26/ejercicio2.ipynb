```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

X = np.array([
    [20, 100],
    [22, 120],
    [23, 140],
    [45, 700],
    [46, 750],
    [48, 720],
    [60, 1500],
    [62, 1600]
])
```

```python
plt.figure(figsize=(6,6))
plt.scatter(X[:,0], X[:,1])

for i in range(len(X)):
    plt.annotate(i+1, (X[i,0], X[i,1]))

plt.title("Clientes")
plt.xlabel("Edad")
plt.ylabel("Gasto")
plt.grid()
plt.show()
```

```python
Z = linkage(X, method='ward')

plt.figure(figsize=(10,5))
dendrogram(Z)
plt.title("Dendrograma Clientes")
plt.show()

modelo2 = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

c2 = modelo2.fit_predict(X)

plt.figure(figsize=(6,6))

for i in range(len(X)):
    plt.scatter(X[i,0], X[i,1], c=f"C{c2[i]}")
    plt.annotate(i+1, (X[i,0], X[i,1]))

plt.title("2 Clusters")
plt.grid()
plt.show()
```

```python
modelo3 = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

c3 = modelo3.fit_predict(X)

plt.figure(figsize=(6,6))

for i in range(len(X)):
    plt.scatter(X[i,0], X[i,1], c=f"C{c3[i]}")
    plt.annotate(i+1, (X[i,0], X[i,1]))

plt.title("3 Clusters")
plt.grid()
plt.show()
```

## ¿Qué clientes parecen más similares?
Los clientes con edades y gastos parecidos.

## ¿Qué grupos naturales identifica el algoritmo?
Clientes de bajo, medio y alto gasto.

## ¿Qué representa cada cluster comercialmente?
Diferentes tipos de clientes.

## ¿Cuál sería una estrategia de marketing para cada grupo?
Descuentos, promociones y beneficios VIP.

## ¿Por qué el algoritmo agrupa ciertos clientes?
Porque tienen características similares.
