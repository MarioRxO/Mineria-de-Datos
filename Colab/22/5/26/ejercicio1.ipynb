```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

X = np.array([
    [1, 1],   
    [2, 1],   
    [5, 5],   
    [6, 5],   
    [10,10],  
    [11,10]   
])
```

```python
plt.figure(figsize=(6,6))
plt.scatter(X[:,0], X[:,1])

labels = ['A','B','C','D','E','F']

for i, txt in enumerate(labels):
    plt.annotate(txt, (X[i,0], X[i,1]))

plt.title("Datos originales")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()
```

```python
Z = linkage(X, method='ward')
```

```python
plt.figure(figsize=(10,5))
dendrogram(Z, labels=labels)
plt.title("Dendrograma")
plt.xlabel("Puntos")
plt.ylabel("Distancia")
plt.show()
```

```python
modelo3 = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

clusters3 = modelo3.fit_predict(X)

plt.figure(figsize=(6,6))

for i in range(len(X)):
    plt.scatter(X[i,0], X[i,1], c=f"C{clusters3[i]}")
    plt.annotate(labels[i], (X[i,0], X[i,1]))

plt.title("3 Clusters")
plt.grid()
plt.show()

print("Clusters con 3 grupos:")
for l,c in zip(labels, clusters3):
    print(l, "-> Cluster", c)
```

```python
modelo2 = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

clusters2 = modelo2.fit_predict(X)

plt.figure(figsize=(6,6))

for i in range(len(X)):
    plt.scatter(X[i,0], X[i,1], c=f"C{clusters2[i]}")
    plt.annotate(labels[i], (X[i,0], X[i,1]))

plt.title("2 Clusters")
plt.grid()
plt.show()

print("\nClusters con 2 grupos:")
for l,c in zip(labels, clusters2):
    print(l, "-> Cluster", c)
```

## ¿Qué puntos se unen primero y por qué?
A y B, C y D, E y F porque están más cerca.

## ¿Qué sucede al cortar el dendrograma en una altura pequeña?
Se forman muchos grupos pequeños.

## ¿Qué cambia al cortar el árbol más arriba?
Los grupos se unen en clusters más grandes.

## ¿Qué significa la altura donde se unen dos ramas?
La distancia entre los grupos.

## ¿Cuántos clusters aparecen al cortar aproximadamente en distancia 8?
Aproximadamente 2 clusters.
