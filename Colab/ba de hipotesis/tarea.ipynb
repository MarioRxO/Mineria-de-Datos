Ejercicio 1
H₀: μ = 500 ml
H₁: μ ≠ 500 ml

```python
import numpy as np
from scipy import stats

data = np.array([498, 501, 499, 502, 500, 497, 503, 499, 501, 500])

t_stat, p_value = stats.ttest_1samp(data, 500)

print("t =", t_stat)
print("p-value =", p_value)
```

Ejercicio 2
H₀: μ₁ = μ₂
H₁: μ₁ ≠ μ₂ 

```python
import numpy as np
from scipy import stats

musica = np.array([65, 70, 68, 72, 66, 69, 71, 67, 70, 68])
silencio = np.array([85, 88, 90, 87, 92, 86, 89, 91, 88, 90])

t_stat, p_value = stats.ttest_ind(musica, silencio, equal_var=False)

print("t =", t_stat)
print("p-value =", p_value)
```

Ejercicio 3
H₀: La distribución sigue las proporciones históricas 
H₁: La distribución es diferente

```python
import numpy as np
from scipy.stats import chisquare

observados = np.array([200, 120, 80])

n = 400
esperados = np.array([0.40, 0.35, 0.25]) * n

chi_stat, p_value = chisquare(f_obs=observados, f_exp=esperados)

print("Chi =", chi_stat)
print("p-value =", p_value)
```
