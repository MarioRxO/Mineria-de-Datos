```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder

# Datos
datos_estres = {
    "Sueño": ["Poco","Poco","Poco","Suficiente","Suficiente","Suficiente","Poco","Suficiente"],
    "Ejercicio": ["No","No","Sí","No","Sí","Sí","Sí","No"],
    "Carga_Tareas": ["Alta","Normal","Alta","Alta","Normal","Alta","Normal","Normal"],
    "Estrés": ["Sí","Sí","Sí","Sí","No","No","No","No"]
}

df_estres = pd.DataFrame(datos_estres)
```

```python
encoders = {}
for col in df_estres.columns:
    le = LabelEncoder()
    df_estres[col] = le.fit_transform(df_estres[col])
    encoders[col] = le

X_estres = df_estres[["Sueño","Ejercicio","Carga_Tareas"]]
y_estres = df_estres["Estrés"]

# Árbol ID3 (Entropy)
modelo_estres = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

modelo_estres.fit(X_estres, y_estres)
```

```python
plt.figure(figsize=(10,6))
plot_tree(
    modelo_estres,
    feature_names=X_estres.columns,
    class_names=["No","Sí"],
    filled=True
)
plt.title("Árbol de Decisión - Nivel de Estrés")
plt.show()
```

```python
nuevo_estudiante = pd.DataFrame({
    "Sueño": ["Poco"],
    "Ejercicio": ["No"],
    "Carga_Tareas": ["Alta"]
})

for col in nuevo_estudiante.columns:
    nuevo_estudiante[col] = encoders[col].transform(
        nuevo_estudiante[col]
    )

pred = modelo_estres.predict(nuevo_estudiante)

resultado = encoders["Estrés"].inverse_transform(pred)

print("Predicción del estudiante:")
print("Estrés =", resultado[0])
```
