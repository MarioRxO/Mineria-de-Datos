```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
```

```python
datos_diabetes = {
    "IMC":["Alto","Alto","Alto","Normal","Normal",
           "Normal","Alto","Normal","Normal","Alto"],

    "Azucar_Sangre":["Alta","Alta","Normal","Alta","Normal",
                     "Normal","Normal","Alta","Normal","Alta"],

    "Actividad_Fisica":["Poca","Mucha","Poca","Poca","Mucha",
                        "Poca","Mucha","Mucha","Poca","Poca"],

    "Diabetes":["Sí","Sí","Sí","Sí","No",
                "No","No","No","No","Sí"]
}

df_diabetes = pd.DataFrame(datos_diabetes)

df_diabetes
```

```python
encoders2 = {}

for col in df_diabetes.columns:
    le = LabelEncoder()
    df_diabetes[col] = le.fit_transform(df_diabetes[col])
    encoders2[col] = le

df_diabetes
```

```python
X_diabetes = df_diabetes[
    ["IMC","Azucar_Sangre","Actividad_Fisica"]
]

y_diabetes = df_diabetes["Diabetes"]
```

```python
id3 = DecisionTreeClassifier(
    criterion="entropy"
)

id3.fit(X_diabetes, y_diabetes)
```

```python
plt.figure(figsize=(12,6))

plot_tree(
    id3,
    feature_names=X_diabetes.columns,
    class_names=["No","Sí"],
    filled=True
)

plt.show()
```

```python
cart = DecisionTreeClassifier(
    criterion="gini"
)

cart.fit(X_diabetes, y_diabetes)
```

```python
plt.figure(figsize=(12,6))

plot_tree(
    cart,
    feature_names=X_diabetes.columns,
    class_names=["No","Sí"],
    filled=True
)

plt.show()
```

```python
paciente = pd.DataFrame({
    "IMC":["Alto"],
    "Azucar_Sangre":["Alta"],
    "Actividad_Fisica":["Poca"]
})

for col in paciente.columns:
    paciente[col] = encoders2[col].transform(
        paciente[col]
    )

pred = id3.predict(paciente)

resultado = encoders2["Diabetes"].inverse_transform(pred)

print("Resultado:", resultado[0])
```
