```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df2 = pd.read_csv('EdadSeguros.csv')
df2.columns = df2.columns.str.strip()
X2 = df2[['Edad']]
y2 = df2['Costo']
```

```python
modelo2 = LinearRegression()
modelo2.fit(X2, y2)
```

```python
edades_clave = np.array([30, 40, 50, 60]).reshape(-1, 1)
predicciones = modelo2.predict(edades_clave)

tabla_pred = pd.DataFrame({
    'Edad': edades_clave.flatten(),
    'Costo Estimado': predicciones
})
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X2, y2, color='darkorange', label='Clientes')
plt.plot(X2, modelo2.predict(X2), color='black', label='Tendencia de Costo')
plt.title('Relación Edad vs. Costo de Póliza')
plt.xlabel('Edad')
plt.ylabel('Costo Mensual')
plt.legend()
plt.show()

print("Parámetros: Pendiente =", modelo2.coef_[0], "Intercepto =", modelo2.intercept_)
print("\nTabla de Predicciones:")
print(tabla_pred)
```
