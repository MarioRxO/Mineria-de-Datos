```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df3 = pd.read_csv('PracticaExamenes.csv')
X3 = df3[['horas']]
y3 = df3['puntuacion']
```

```python
modelo3 = LinearRegression()
modelo3.fit(X3, y3)
```

```python
m = modelo3.coef_[0]
b = modelo3.intercept_
r_cuadrado = r2_score(y3, modelo3.predict(X3))
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X3, y3, color='seagreen', alpha=0.5)
plt.plot(X3, modelo3.predict(X3), color='red')
plt.title('Horas de Práctica vs. Puntuación Final')
plt.xlabel('Horas Semanales')
plt.ylabel('Puntuación')
plt.show()

print(f"Ecuación de la recta: Puntuación = {m:.2f} * Horas + {b:.2f}")
print(f"Coeficiente de determinación (R²): {r_cuadrado:.4f}")
print("Interpretación: El R² indica qué tanto de la variación en la nota se explica por las horas de estudio.")
```

```python
horas_input = 7
print(f"Predicción para {horas_input} horas: {modelo3.predict([[horas_input]])[0]:.2f}")
```
