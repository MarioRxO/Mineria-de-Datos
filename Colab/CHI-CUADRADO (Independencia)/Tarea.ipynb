```python
import numpy as np
from scipy.stats import chi2_contingency
```

```python
#Ejercicio 1
tabla1 = np.array([
    [85, 65],
    [50, 100]
])

# Prueba Chi-cuadrada
chi2, p, dof, expected = chi2_contingency(tabla1)

print("\nTabla observada:\n", tabla1)
print("\nFrecuencias esperadas:\n", expected)
print("\nChi-cuadrada:", chi2)
print("Grados de libertad:", dof)
print("p-valor:", p)

# Interpretación
alpha = 0.05
if p < alpha:
    print("\nConclusión: Existe relación significativa entre ciudad y transporte")
else:
    print("\nConclusión: No existe relación significativa")
```
