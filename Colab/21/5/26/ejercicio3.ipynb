```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
import numpy as np

# 1 DataFrame
df3 = pd.read_csv("ejercicio_3_fraude_bancario.csv")
```

```python
# 2 Escalar los datos
scaler = StandardScaler()
X3 = scaler.fit_transform(df3[["Monto", "Hora"]])
```

```python
# 3 k-distancia
k = 3
nn = NearestNeighbors(n_neighbors=k)
nn.fit(X3)
distancias, indices = nn.kneighbors(X3)
distancias = np.sort(distancias[:, k-1])

plt.figure(figsize=(6, 5))
plt.plot(distancias)
plt.title("Gráfico k-distancia")
plt.xlabel("Puntos ordenados")
plt.ylabel("Distancia al k-ésimo vecino")
plt.grid(True)
plt.show()
```

```python
# 4  DBSCAN
eps = 1.0  # ajusta según la gráfica
db3 = DBSCAN(eps=eps, min_samples=3)
df3["Cluster"] = db3.fit_predict(X3)


df3["Estado"] = df3["Cluster"].apply(lambda x: "Sospechosa" if x == -1 else "Normal")
```

```python
# 5 Graficar
plt.figure(figsize=(7, 5))

for c in df3["Cluster"].unique():
    subset = df3[df3["Cluster"] == c]
    if c == -1:
        plt.scatter(subset["Monto"], subset["Hora"], c="black", label="Fraude / Sospechosa", marker="x", s=100)
    else:
        plt.scatter(subset["Monto"], subset["Hora"], label=f"Cluster {c}")

plt.title("Transacciones bancarias detectadas con DBSCAN")
plt.xlabel("Monto")
plt.ylabel("Hora")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 6 Porcentaje 
porcentaje = (df3["Cluster"].eq(-1).sum() / len(df3)) * 100
print("Porcentaje de transacciones sospechosas:", round(porcentaje, 2), "%")
print(df3)
```

DBSCAN tiene ventaja sobre K-Means para fraude porque no necesita fijar el número de grupos y además puede detectar transacciones aisladas como ruido.
